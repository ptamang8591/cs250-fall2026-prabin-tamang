## Section 3 · Instance Methods
**About 5 minutes · 15 points**

A **method** is a function defined inside a class. It uses `self` to read and change the object’s own attributes.

**Your task:** complete `WaterBottle`.
- `WaterBottle(capacity)` stores `capacity`; the bottle starts empty (`amount = 0`).
- `fill(ml)` adds `ml`, but `amount` never goes above `capacity`.
- `drink(ml)` removes `ml`, but `amount` never goes below `0`. It **returns** how much was actually drunk.
- `is_empty()` returns `True` when `amount` is `0`.

```python
class WaterBottle:

    def __init__(self, capacity):
        # TODO: store capacity; the bottle starts empty
        self.capacity = capacity
        self.amount = 0

    def fill(self, ml):
        # TODO: add ml, but never go above capacity
        self.amount = min(self.capacity, self.amount + ml)

    def drink(self, ml):
        # TODO: remove ml, but never go below 0
        #       and return how much was actually drunk
        actual_drunk = min(self.amount, ml)
        self.amount -= actual_drunk
        return actual_drunk

    def is_empty(self):
        # TODO: return True if the bottle is empty
        return self.amount == 0
```

**Check your work:** run the cell below.

```python
def _fill_caps():
    b = WaterBottle(500); b.fill(300); b.fill(300)
    return b.amount == 500

def _drink_some():
    b = WaterBottle(500); b.fill(300)
    return b.drink(200) == 200 and b.amount == 100

def _drink_too_much():
    b = WaterBottle(500); b.fill(300)
    return b.drink(1000) == 300 and b.amount == 0

run_checks("3 · Instance Methods", [
    ("A new bottle is empty", lambda: WaterBottle(500).amount == 0 and WaterBottle(500).capacity == 500,
     "In __init__: self.capacity = capacity and self.amount = 0"),
    ("is_empty() is True for a new bottle", lambda: WaterBottle(500).is_empty() is True,
     "return self.amount == 0"),
    ("fill() never goes above capacity", _fill_caps,
     "min(self.amount + ml, self.capacity) keeps it at the cap."),
    ("drink() removes water and returns the amount drunk", _drink_some,
     "Remember to return the amount at the end of drink()."),
    ("drink() never goes below 0", _drink_too_much,
     "You can only drink what is in the bottle: min(ml, self.amount)."),
])
```

## Section 4 · Class & Static Methods
**About 4 minutes · 15 points**

- A **static method** (`@staticmethod`) gets no `self`: it is a helper that belongs with the class.
- A **class method** (`@classmethod`) gets the class as `cls`: it is often another way to build an object.

**Your task:** complete `Color`.
- `Color.is_valid(value)` returns `True` if `value` is an `int` from 0 to 255. Make it a **static method**.
- `Color.from_hex("#FF8800")` returns `Color(255, 136, 0)`. Make it a **class method**. Hint: `int("FF", 16)` is `255`.
- `Color(r, g, b)` raises `ValueError` if any channel is not valid (use `Color.is_valid`).

```python
class Color:

    def __init__(self, r, g, b):
        # TODO: raise ValueError if any of r, g, b is not valid (use Color.is_valid)
        #       otherwise store r, g and b
        if (
            not Color.is_valid(r)
            or not Color.is_valid(g)
            or not Color.is_valid(b)
        ):
            raise ValueError("RGB values must be integers between 0 and 255.")
        self.r = r
        self.g = g
        self.b = b

    # TODO: add the right decorator
    @staticmethod
    def is_valid(value):
        # TODO: True if value is an int from 0 to 255
        return isinstance(value, int) and 0 <= value <= 255

    # TODO: add the right decorator
    @classmethod
    def from_hex(cls, hex_string):
        # TODO: "#FF8800" -> cls(255, 136, 0)
        #       hex_string[1:3] is "FF", and int("FF", 16) is 255
        r = int(hex_string[1:3], 16)
        g = int(hex_string[3:5], 16)
        b = int(hex_string[5:7], 16)
        return cls(r, g, b)
```

## Section 5 · Dunder Methods
**About 5 minutes · 15 points**

Dunder (double underscore) methods let your objects work with Python’s built-in syntax. Python calls them for you.

**Your task:** add three dunder methods to `Movie`.
- `__str__`: `print(Movie("Inception", 2010, 8.8))` shows `Inception (2010) - 8.8/10`
- `__eq__`: two movies are equal when their **title and year** match (rating can differ)
- `__lt__`: compare by **rating**, so `sorted(movies)` goes from lowest to highest rating

```python
class Movie:

    def __init__(self, title, year, rating):
        self.title = title
        self.year = year
        self.rating = rating

    # TODO 5a: __str__  ->  "Inception (2010) - 8.8/10"
    def __str__(self):
        return f"{self.title} ({self.year}) - {self.rating}/10"

    # TODO 5b: __eq__   ->  equal when title AND year match
    def __eq__(self, other):
        if not isinstance(other, Movie):
            return False
        return self.title == other.title and self.year == other.year

    # TODO 5c: __lt__   ->  compare by rating
    def __lt__(self, other):
        return self.rating < other.rating
```

```python
# Try it: run this once your class works (it shows an error until then)
films = [Movie("Up", 2009, 8.3), Movie("Inception", 2010, 8.8), Movie("Cars", 2006, 7.2)]
print(films[0])
for f in sorted(films):
    print(f)
```

## Section 6 · Inheritance & super()
**About 4 minutes · 15 points**

A **child class** gets every attribute and method of its **parent**. It can **override** a method, and call the parent’s version with `super()`.

Run the `MenuItem` cell below (do not change it). Then write `Drink(MenuItem)`:
- `Drink(name, price, size)`: use `super().__init__(name, price)` for name and price, then store `size` (`"S"`, `"M"` or `"L"`).
- Override `label()` so it includes the size and adds the size extra from `EXTRA`:
  `Drink("Latte", 4.50, "L").label()` → `Latte (L): $5.50`

```python
# Given parent class: run this cell, do not change it
class MenuItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def label(self):
        return f"{self.name}: ${self.price:.2f}"

print(MenuItem("Bagel", 3).label())
```

```python
# Given parent class
class MenuItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def label(self):
        return f"{self.name}: ${self.price:.2f}"


# Child class
class Drink(MenuItem):  # TODO 6a: make Drink inherit from MenuItem
    EXTRA = {"S": 0.00, "M": 0.50, "L": 1.00}

    def __init__(self, name, price, size):
        # TODO 6b: call the parent's __init__ with super() to set name and price
        super().__init__(name, price)
        # TODO 6c: store size
        self.size = size

    def label(self):
        # TODO 6d: return "Latte (L): $5.50"  (price + the size extra, 2 decimals)
        total_price = self.price + self.EXTRA[self.size]
        return f"{self.name} ({self.size}): ${total_price:.2f}"


# Test check:
latte = Drink("Latte", 4.50, "L")
print(latte.label())  # Output: Latte (L): $5.50
```

## Section 7 · Polymorphism
**About 3 minutes · 5 points**

**Polymorphism:** different classes answer the same method call in their own way. The loop below never checks which class an alert is.

**Your task:** write two classes, each with a `send(message)` method that **returns** a string:
- `EmailAlert("ana@school.edu").send("Hi")` → `Email to ana@school.edu: Hi`
- `SMSAlert("555-0100").send("Hi")` → `SMS to 555-0100: Hi`, but SMS messages are cut to their **first 20 characters**

```python
class EmailAlert:

    def __init__(self, email):
        self.email = email

    def send(self, message):
        return f"Email to {self.email}: {message}"


class SMSAlert:

    def __init__(self, phone_number):
        self.phone_number = phone_number

    def send(self, message):
        # Truncate message to a maximum of 20 characters
        truncated_message = message[:20]
        return f"SMS to {self.phone_number}: {truncated_message}"
```

```python
# Try it: the same loop works for both classes (it shows an error until they exist)
alerts = [EmailAlert("ana@school.edu"), SMSAlert("555-0100")]
for a in alerts:
    print(a.send("Your lab is due Friday at midnight"))
```

## Section 8 · Encapsulation with @property
**About 3 minutes · 10 points**

**Encapsulation:** the object protects its own rules. A `@property` looks like a normal attribute from outside, but a method checks every new value.

**Your task:** complete `Thermostat`.
- The temperature is stored in `self._temp` (the underscore means “internal”).
- `t.temp` returns it.
- Setting `t.temp = value` raises `ValueError` if `value` is below 10 or above 30.
- `__init__` must set the starting value **through the property**, so `Thermostat(5)` is rejected too.

```python
class Thermostat:

    def __init__(self, temp):
        # TODO 8a: set the starting value through the property: self.temp = temp
        self.temp = temp

    @property
    def temp(self):
        # TODO 8b: return self._temp
        return self._temp

    @temp.setter
    def temp(self, value):
        # TODO 8c: raise ValueError if value < 10 or value > 30
        #          otherwise store it in self._temp
        if value < 10 or value > 30:
            raise ValueError("Temperature must be between 10 and 30.")
        self._temp = value
```

**Check your work:** run the cell below.

```python
def _set_ok():
    t = Thermostat(21); t.temp = 25
    return t.temp == 25

def _set_bad():
    t = Thermostat(21)
    return raises(ValueError, lambda: setattr(t, "temp", 35)) and t.temp == 21

run_checks("8 · Encapsulation with @property", [
    ("Thermostat(21).temp is 21, through a property", lambda: isinstance(inspect.getattr_static(Thermostat, "temp"), property) and Thermostat(21).temp == 21,
     "__init__: self.temp = temp; getter: return self._temp"),
    ("The value is stored in _temp", lambda: vars(Thermostat(21)).get("_temp") == 21,
     "The setter should store self._temp = value."),
    ("Setting a valid value works", _set_ok,
     "The setter stores the value when it is in range."),
    ("Setting 35 raises ValueError and keeps 21", _set_bad,
     "Raise before storing: if value < 10 or value > 30: raise ValueError(...)"),
    ("Thermostat(5) raises ValueError", lambda: raises(ValueError, lambda: Thermostat(5)),
     "In __init__, write self.temp = temp (no underscore) so the setter runs."),
])
```

## Bonus · Composition (not graded)
**If you finish early**

**Composition** is a “has-a” relationship: a `Garden` *has* plants; it is not a kind of plant. It stores `Plant` objects from Section 1 in a list.

Write `Garden` with:
- `add(plant)`: adds a `Plant` object
- `thirsty()`: returns a list of the **names** of plants with `water_level` below 20
- `__len__`: so `len(garden)` is the number of plants

```python
class Garden:

    def __init__(self):
        self.plants = []

    def add(self, plant):
        """Adds a Plant object to the garden."""
        self.plants.append(plant)

    def thirsty(self):
        """Returns a list of plant names with water_level below 20."""
        return [
            plant.name for plant in self.plants if plant.water_level < 20
        ]

    def __len__(self):
        """Returns the total number of plants in the garden."""
        return len(self.plants)
```
